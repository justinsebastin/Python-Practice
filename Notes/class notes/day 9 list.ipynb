```python
fruits = ["apple","orange","mango"]
vegetable = ["tomato", "onion", ]
fruits.append("grapes") #add element at end
fruits.insert(1,"guava") #add element at index
fruits.extend(vegetable) #add another build in list to the list
fruits.extend(["bringal","pumkin"]) #add another list to the existing list
fruits[1] = "water melon" #update the list
fruits.pop() #remove last
fruits.pop(2) #remove the indexed item in list
fruits.remove("water melon") #remove the element in by item number
fruits.reverse() #reverse the element in the list
print(fruits.count("apple")) #to count the element the list
print(fruits.index("tomato")) #print the index value of the element
print(fruits)
```
