slicing

```python
num = [10,20,30,40,40,50,60,70,80,90]
print(num[1:3])
```

```python
print(num[0:10:2])
```

```python
print(num[:4])
```

```python
print(num[4::2])
```

```python
print(num[::-1])
```

```python
print(num[3:4])
```

```python
print(num[3])
```

```python
print(num[2:2])
```

```python
print(num[15:])
```

```python
print(num[-10:-20])
```

```python
print(num[1:3:0])
```

```python
num[1:4] = [100,200]
print(num)
```
