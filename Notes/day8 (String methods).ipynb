```python
data = "Good Morning did you have you're coffee."

l = []
temp = ""
for i in data:
  if i == " " or i == ".":
      l.append(temp)
      temp = ""
  else:
      temp += i


print(l)
```

```python
# 1. Split using spaces
data1 = "Python is easy to learn"
print("Split using spaces:", data1.split())


# 2. Split using commas
data2 = "Apple,Banana,Orange,Mango"
print("Split using comma:", data2.split(','))


# 3. Split using hyphens
data3 = "2026-11-20"
print("Split using hyphen:", data3.split('-'))


# 4. Split using colons
data4 = "10:30:45"
print("Split using colon:", data4.split(':'))


# 5. Split using underscores
data5 = "my_python_learning_journey"
print("Split using underscore:", data5.split('_'))


# 6. Split using dots
data6 = "www.google.com"
print("Split using dot:", data6.split('.'))


# 7. Split using forward slashes
data7 = "home/santosh/documents/python"
print("Split using slash:", data7.split('/'))


# 8. Split using multiple spaces
data8 = "Python    is    very    easy"
print("Split using multiple spaces:", data8.split())


# 9. Split using newline
data9 = "Apple\nBanana\nOrange\nMango"
print("Split using newline:", data9.split('\n'))


# 10. Split using maximum number of splits
data10 = "Apple,Banana,Orange,Mango"
print("Split using maxsplit:", data10.split(',', 2))
```
