# Python Notes – List Slicing

List slicing is a technique in Python used to extract a portion of elements from a list.

It allows us to access multiple elements from a list using their index positions without accessing each element separately.

The original list remains unchanged when we perform normal slicing.

## 1. What is List Slicing?

**Definition:**

List slicing is a technique in Python used to extract a portion of elements from a list. It allows us to access multiple elements from a list using their index positions without accessing each element separately.

**Example:**

````python
numbers = [10, 20, 30, 40, 50, 60]

result = numbers[1:4]

print(result)
````

**Output:**

```text
[20, 30, 40]
```

**Explanation:**

The list contains the following elements:

| Element | Index |
|---|---:|
| 10 | 0 |
| 20 | 1 |
| 30 | 2 |
| 40 | 3 |
| 50 | 4 |
| 60 | 5 |

In `numbers[1:4]`:

- Starting index is 1.
- Ending index is 4.
- Python extracts elements from index 1 up to index 3.
- Index 4 is not included.

**Important rule:** The starting index is included, but the ending index is excluded.

`numbers[1:4]` → `[20, 30, 40]`

## 2. Syntax of List Slicing

The general syntax is:

```python
list_name[start : stop : step]
```

There are three parts:

| Parameter | Meaning |
|---|---|
| start | The index where slicing begins |
| stop | The index where slicing ends (excluded) |
| step | The number of positions to move each time |

**Example:**

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:5:2])
````

**Output:**

```text
[20, 40]
```

**Explanation:**

- Start at index 1 → 20.
- Move 2 positions → 40.
- Stop before index 5.

Remember: `start` and `stop` are positions, whereas `step` controls how many positions Python skips.

## 3. Different Working Scenarios of List Slicing

We will use the following list for all the examples in this section.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]
````

### Example 1: Slicing with Starting and Ending Index

Get elements from index 1 to index 4.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[1:4])
````

**Output:**

```text
[20, 30, 40]
```

**Explanation:** Index 1 is included, and index 4 is excluded.

### Example 2: Slicing Without Starting Index

What happens when we do not specify the starting index?

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[:4])
````

**Output:**

```text
[10, 20, 30, 40]
```

**Explanation:** When the starting index is not provided, Python automatically starts from index 0.

`numbers[:4]` means start from the beginning and stop before index 4.

### Example 3: Slicing Without Ending Index

What happens when we do not specify the ending index?

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[3:])
````

**Output:**

```text
[40, 50, 60, 70]
```

**Explanation:** When the ending index is not provided, Python continues until the last element.

### Example 4: Slicing Without Start and Stop

What happens when both starting and ending indexes are omitted?

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[:])
````

**Output:**

```text
[10, 20, 30, 40, 50, 60, 70]
```

**Explanation:** Python returns all elements from the beginning to the end of the list.

This also creates a new outer list instead of returning the original list itself.

### Example 5: Slicing Using Step

We can skip elements using the step parameter.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[::2])
````

**Output:**

```text
[10, 30, 50, 70]
```

**Explanation:**

- Start from index 0.
- Move 2 positions each time.
- Select indexes 0, 2, 4 and 6.

The default step is 1.

### Example 6: Slicing with Starting Index and Step

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[1::2])
````

**Output:**

```text
[20, 40, 60]
```

**Explanation:**

- Start from index 1.
- Move 2 positions each time.
- Continue until the end of the list.

### Example 7: Slicing with Negative Indexing

Python also supports negative indexes. Negative indexing starts from the end of the list.

| Element | Positive Index | Negative Index |
|---|---:|---:|
| 10 | 0 | -7 |
| 20 | 1 | -6 |
| 30 | 2 | -5 |
| 40 | 3 | -4 |
| 50 | 4 | -3 |
| 60 | 5 | -2 |
| 70 | 6 | -1 |

**Example:**

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[-4:-1])
````

**Output:**

```text
[40, 50, 60]
```

**Explanation:**

- Start at index -4 → 40.
- Stop before index -1 → 70.
- Therefore, the result is `[40, 50, 60]`.

### Example 8: Slicing Using Negative Step

A negative step moves through the list in reverse order.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[::-1])
````

**Output:**

```text
[70, 60, 50, 40, 30, 20, 10]
```

**Explanation:**

- The step is -1.
- Python starts from the last element.
- It moves backwards one position at a time.

This is one of the easiest ways to reverse a list.

### Example 9: Reverse a Specific Portion of a List

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[5:1:-1])
````

**Output:**

```text
[60, 50, 40, 30]
```

**Explanation:**

- Start at index 5 → 60.
- Move backwards by 1.
- Stop before index 1.

The selected indexes are 5, 4, 3 and 2.

### Example 10: Extracting a Single Element Using Slicing

We can also extract just one element using slicing.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[2:3])
````

**Output:**

```text
[30]
```

**Explanation:**

- Starts at index 2.
- Stops before index 3.
- Returns a list containing only 30.

**Important difference:**

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[2])
````

**Output:**

```text
30
```

The first returns an integer, while slicing returns a list.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[2:3])
````

**Output:**

```text
[30]
```

## 4. Understanding Empty Slices

Sometimes slicing returns an empty list. This is not necessarily an error.

### Example 11: Starting Index Greater Than Ending Index

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[4:2])
````

**Output:**

```text
[]
```

**Explanation:** The default step is positive 1. Python cannot move forward from index 4 to reach index 2, so it returns an empty list.

To move backwards, use a negative step:

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[4:2:-1])
````

**Output:**

```text
[50, 40]
```

### Example 12: Starting and Ending at the Same Index

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[2:2])
````

**Output:**

```text
[]
```

**Explanation:** The starting index and ending index are the same. Since the ending index is excluded, no elements are selected.

## 5. Out-of-Range Index Scenarios

An important difference between normal indexing and slicing is how they handle indexes outside the list.

### Example 13: Ending Index Greater Than List Length

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:10])
````

**Output:**

```text
[20, 30, 40, 50]
```

**Explanation:** The list has only five elements, but we requested an ending index of 10. Python automatically stops at the end of the list. This does not generate an error.

### Example 14: Starting Index Greater Than List Length

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[10:15])
````

**Output:**

```text
[]
```

**Explanation:** The starting index is beyond the list length. Therefore, no elements are available to extract. Again, this does not generate an error.

### Example 15: Negative Index Beyond List Length

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[-10:3])
````

**Output:**

```text
[10, 20, 30]
```

**Explanation:** A negative starting position beyond the length is clipped to the beginning of the list. Python still performs slicing safely.

## 6. Error Scenarios in List Slicing

Normal slicing is flexible, but certain invalid values can generate errors.

### Error 1: Using Float as an Index

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1.5:4])
````

**Output (error):**

```text
TypeError: slice indices must be integers or None or have an __index__ method
```

**Why does this error occur?**

Python list slicing requires integer indexes. A decimal value such as `1.5` cannot be used as a slice index.

**Correct code:**

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
````

**Output:**

```text
[20, 30, 40]
```

### Error 2: Using Zero as Step

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[::0])
````

**Output (error):**

```text
ValueError: slice step cannot be zero
```

**Why does this error occur?**

The step determines how many positions Python should move. A step of zero means Python cannot move forward or backward. Therefore, zero is not allowed.

**Correct code:**

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[::1])
````

**Output:**

```text
[10, 20, 30, 40, 50]
```

### Error 3: Slicing a Non-Iterable Number

````python
num = 100

print(num[1:3])
````

**Output (error):**

```text
TypeError: 'int' object is not subscriptable
```

**Why does this error occur?**

An integer is not a list, so we cannot perform list slicing on it.

**Correct code:**

````python
numbers = [100, 200, 300, 400]

print(numbers[1:3])
````

**Output:**

```text
[200, 300]
```

### Error 4: Using a String as a Slice Index

````python
numbers = [10, 20, 30, 40, 50]

print(numbers["1":"4"])
````

**Output (error):**

```text
TypeError
```

**Why does this error occur?**

The values `"1"` and `"4"` are strings, not integers.

**Correct code:**

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
````

**Output:**

```text
[20, 30, 40]
```

## 7. Difference Between Indexing and Slicing

This is an important concept to understand.

| Indexing | Slicing |
|---|---|
| Accesses a single element | Extracts multiple elements |
| Uses one index | Uses start, stop and step |
| Returns the element itself | Returns a new list |
| Can generate `IndexError` for an invalid index | Out-of-range slice boundaries are normally handled safely |

**Example:**

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[2])
````

**Output:**

```text
30
```

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[2:4])
````

**Output:**

```text
[30, 40]
```

## 8. Modifying a List Using Slicing

Slicing can also be used to replace or delete portions of a list.

### Example 16: Replacing Elements

````python
numbers = [10, 20, 30, 40, 50]

numbers[1:3] = [200, 300]

print(numbers)
````

**Output:**

```text
[10, 200, 300, 40, 50]
```

**Explanation:** The elements at indexes 1 and 2 are replaced with 200 and 300.

### Example 17: Deleting Elements

````python
numbers = [10, 20, 30, 40, 50]

del numbers[1:3]

print(numbers)
````

**Output:**

```text
[10, 40, 50]
```

**Explanation:** The elements at indexes 1 and 2 are deleted from the original list.

## 9. Important Rules to Remember

| Rule | Example | Result |
|---|---|---|
| Start is included | `a[1:4]` | Starts at index 1 |
| Stop is excluded | `a[1:4]` | Stops before index 4 |
| Default start is 0 | `a[:3]` | First 3 elements |
| Default stop is the end | `a[2:]` | From index 2 to last |
| Default step is 1 | `a[::]` | All elements |
| Positive step | `a[::2]` | Every second element |
| Negative index | `a[-3:]` | Last 3 elements |
| Negative step | `a[::-1]` | Reverses the list |
| Zero step | `a[::0]` | `ValueError` |
| Float index | `a[1.5:3]` | `TypeError` |
| Out-of-range boundary | `a[1:100]` | Boundary is clipped to list length |
