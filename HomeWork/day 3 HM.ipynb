### Positive or Negative

```python
a = int(input("Enter the number"))
if a < 0:
    print("Its is negative number")
else:
    print("Its is positive number")
```

### Pass or Fail

```python
mark = int(input("Enter the mark"))
if mark >= 35:
    print("pass")
else:
    print("Fail")
```

### Voting Eligibility

```python
age = int(input("Enter the age"))
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

### Bigger Number

```python
num_1 = int(input("Enter the number 1"))
num_2 = int(input("Enter the number 2"))
print(f"among {num_1} and {num_2}")
if num_1>num_2:
    print(f"{num_1} is greater")
else:
    print(f"{num_2} is greater")
```

### Divisible by 5
