```python
#charecter analysis
g = input("enter the word")
h = list(g)
# print(h)
alp = 0
num = 0
spa = 0
special = 0
for i in h:
    # print(i)
     if i.isalpha() == True:
         alp = alp + 1
     elif i.isdigit() == True:
         num = num+1
     elif i.isspace() == True:
         spa = spa + 1
     else:
         special = special+1
print("Alphabet",alp)  
print("Digits:",num)
print("Spaces:",spa)
print("Special Characters:",special) 
```
