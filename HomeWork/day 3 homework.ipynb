### Grade Calculator

```python
mark_1 = int(input("Enter the mark"))
if mark_1 >= 90:
    print("A grade")
elif mark_1 >= 75:
    print("B grade")
elif mark_1 >= 50:
    print("C grade")
else:
    print("FAIL")
```

### ATM Withdrawal

```python
amount = 10000
wd_amount = int(input("Enter the amount to be withdraw"))
if wd_amount <= amount:
    print("Transaction Sucessfull")
else:
    print("Insufficient Balance")
```

### Scholarship Eligibility (Nested If)

```python
mark_2 = int(input("enter the secured mark"))
school = input("school gov/private")
if mark_2 >= 70:
    if school == "gov":
        print("Eligible for scholarship")
    else:
        print("Not Eligible")
else:
    print("not eligible")
```
