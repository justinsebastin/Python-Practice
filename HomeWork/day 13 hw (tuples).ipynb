```python
# 1. Add an Element to a Tuple using Type Conversion
t = (10,20,30,40)
a = list(t)
a.append(50)
t = tuple(a)
print(t)
```

```python
# 2. Update a Value in a Tuple using Type Conversion
student = ("Dhana", 21, "Chennai")
a = list(student)
b = a.index("Chennai")
a[b] = "Vellore"
student = tuple(a)
print(student)
```

```python
# 3. Store Fixed Product Details as Tuple
a = ("Laptop", 55000, "Electronics", 4.5)
print("Product name:", a[0])
print("Price:", a[1])
print("Category:", a[2])
print("Rating:", a[3])
```

```python
# 4. Swap First and Last Element of a Tuple
t = (10, 20, 30, 40, 50)
a = list(t)
a[0], a[-1] = a[-1], a[0]
t = tuple(a)
print(t)
```

```python
# 5. Count Frequency of a Value in Tuple
marks = (78, 82, 78, 90, 67, 78, 82)
for i in set(marks):
    print(f"Frequency of {i} is: {marks.count(i)}")
```

```python
# 6. Store Dataset Shape as Tuple
shape = (500, 12)

print("Dataset Shape:", shape)
print("Rows:", shape[0])
print("Columns:", shape[1])
```

```python
# 7. Feature Names as Tuple
feature_names = ("age", "gender", "income", "education", "occupation")
print("Feature Names:", feature_names)
print("Total feature",len(feature_names))
if "income" in feature_names:
    print("Income is present in the feature names.")
else:
    print("Income is not presented")
```

```python
# 8. Train/Test Split Info as Tuple
t = (800, 200)
print("Training size",t[0])
print("Testing size",t[1])
print("Total samples",int(t[0])+int(t[1]))
```

```python
# 9. RGB Pixel as Tuple
c = (120, 200, 150)
print(f"Red = {c[0]}, Green = {c[1]}, Blue = {c[2]}")
print("Average intensity",sum(c)/len(c))
```

```python
# 10. Model Evaluation Scores Tuple
scores = (0.91, 0.88, 0.90)
print("Accuracy:", scores[0])
print("Precision:", scores[1])
print("Recall:", scores[2])
print("Average Score:", sum(scores)/len(scores))
```
