```python
#number odd and even number in list
a = [12,32,34,55,34,54,66,54,55,65,75,43]
even = 0
odd = 0
for i in a:
    if i%2 == 0:
        even += 1
    else:
        odd += 1
print("even-",even)
print("odd-",odd)
```

```python
#remove dublicate
a = [12,32,43,23,43,53,33,12,32,33,53,53]
b = []
for i in a:
    if i not in b:
        b.append(i)
print(b)
```

```python
#find missing number in sequence
a = [1,2,4,5,6]
for i in range(1,len(a)):
    if i not in a:
        print("missing number",i)
```

```python
# reverse the list 
a = [1,2,3,4,5,6,7,8]
b = a[::-1]
print(b)
```

```python
#move 0 to the end
a = [1,2,0,7,6,0,9,8]
b = []
zeros = []
for i in a:
    if i != 0:
        b.append(i)
for i in a:
    if i == 0:
        zeros.append(i)
c = b + zeros
print(c)
```

```python
#common element
a = [1,2,3,4]
b = [3,4,5,6]
c = []
for i in a:
    for j in b:
        if i == j:
            c.append(j)
print(c)
```

```python
# move to the last
a = [1,2,3,4,5]
b = a[-1]
c  = [b]
for i in range (0,len(a)-1):
    c.append(a[i])
print(c)
```

```python
# finding pair of resulted sum 
a = [1,2,3,4,5]
b = 9
c = []
for i in a:
    for j in a:
        if i+j == b:
            c.append(i)
            c.append(j)
d = list(set(c))
print(d)
```
