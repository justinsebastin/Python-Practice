### add even number

```python
a = int(input("Enter a Number"))
add = 0
for i in range(1,a+1):
    if a%2 ==0:
        add = add+i
print(add)
```

### reverse a number

```python
e = input("Enter the number to reverse")
rev = " "
for digit in e:
    rev = digit + rev
print(rev)
```

### Tables

```python
for i in range (1,11):
    for j in range(1,11):
        print(i,"x",j,"=",i*j,end = "   ")
    print()
```

### Factorial

```python
num_2 = int(input("Enter the number"))
fact = 1
for i in range(1,num_2+1):
    fact = fact*i
print(fact)
```

### Largest Number in list

```python
list = [22,33,42,34,54,23,55,67,54,33,22]
large = list[0]
for i in list:
    if i > large:
        large = i
print(large)
```

### prime number

```python
c = int(input("Enter the number"))
for i in range(2,c):
    if c%i == 0:
        print("Its not prime number")
        break
else:
    print("Its is a prime number")
```
